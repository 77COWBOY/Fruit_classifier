```python
import os
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader

from torch.autograd import Variable
from torchvision import datasets,models
import torchvision.transforms as transforms
import torch.optim as optim
import time

import matplotlib.pyplot as plt
```

```python
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)

```

# Create transformers

```python
train_transform = transforms.Compose([ 
    transforms.Resize((226,226)),
    transforms.ToTensor(),
   transforms.Normalize(mean=[0.485,0.456,0.406],std=[0.229,0.224,0.225])
])
```

```python
val_transform = transforms.Compose([ 
    transforms.Resize((226,226)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485,0.456,0.406],std=[0.229,0.224,0.225])
])
```

```python
train_dir='fruits-360-original-size/Training/'
val_dir='fruits-360-original-size/Validation//'
```

* Read & transform our data

```python
train_data = datasets.ImageFolder(os.path.join(train_dir),transform=train_transform)
val_data = datasets.ImageFolder(os.path.join(val_dir),transform=val_transform)
```

* create dataloaders

```python
train_loader=DataLoader(train_data,batch_size=16,shuffle=True)
val_loader=DataLoader(val_data,batch_size=16,shuffle=True)

class_names= train_data.classes
num_classes= len(class_names)
```

```python
# get image from the dataset (C, H, W)
image, label = train_data[0]

# mean & standar deviation used in transformers
mean = torch.tensor([0.485, 0.456, 0.406]).view(3, 1, 1)
std = torch.tensor([0.229, 0.224, 0.225]).view(3, 1, 1)

# denorm
image_denorm = image * std + mean

image_denorm = torch.clamp(image_denorm, 0, 1)

# order axes(C, H, W) to (H, W, C) just for matplot lib to be able to plot
image_to_show = image_denorm.permute(1, 2, 0)

plt.imshow(image_to_show)
plt.title(f"Label: {label}")
plt.axis('off')
plt.show()
```

```python
model02 = models.efficientnet_b2(
    weights=models.EfficientNet_B2_Weights.DEFAULT
)
num_classses = len(class_names)

model02.classifier[1] = nn.Linear(
    model02.classifier[1].in_features,
    num_classes
)
```

# Define optimizer & loss funtion

```python
optimizer = torch.optim.Adam(model02.parameters(), lr=1e-3)
criterion = nn.CrossEntropyLoss()
```

# training our Model

```python
model02 = model02.to(device)
```

```python
import torch

epochs = 15

train_losses = []
test_losses = []
train_correct = []
test_correct = []

total_train_imgs = len(train_loader.dataset)
total_test_imgs = len(val_loader.dataset)



for i in range(epochs):
    trn_corr = 0
    tst_corr = 0
    running_train_loss = 0.0
    running_test_loss = 0.0
    # training
    model02.train()

    
    model02.train()
    for X_train, y_train in train_loader:
        X_train, y_train = X_train.to(device), y_train.to(device)

        # Forward
        y_pred = model02(X_train)
        loss = criterion(y_pred, y_train)

        
        predicted = torch.max(y_pred, 1)[1]
        trn_corr += (predicted == y_train).sum().item()
        running_train_loss += loss.item() * X_train.size(0)

        # Backward & Step
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()

   
    epoch_train_loss = running_train_loss / total_train_imgs
    epoch_train_acc = (trn_corr / total_train_imgs) * 100
    train_losses.append(epoch_train_loss)
    train_correct.append(epoch_train_acc)

    
    model02.eval()
    with torch.no_grad():
        for X_test, y_test in val_loader:
            X_test, y_test = X_test.to(device), y_test.to(device)

            y_val = model02(X_test)
            loss_v = criterion(y_val, y_test)

            running_test_loss += loss_v.item() * X_test.size(0)
            predicted_val = torch.max(y_val, 1)[1]
            tst_corr += (predicted_val == y_test).sum().item()

   
    epoch_test_loss = running_test_loss / total_test_imgs
    epoch_test_acc = (tst_corr / total_test_imgs) * 100
    test_losses.append(epoch_test_loss)
    test_correct.append(epoch_test_acc)

    print(f'******* End of epoch {i+1} *******')
    print(f'Train Loss: {epoch_train_loss:.4f} | Train Acc: {epoch_train_acc:.2f}%')
    print(f'Test Loss:  {epoch_test_loss:.4f} | Test Acc:  {epoch_test_acc:.2f}%\n')
```
