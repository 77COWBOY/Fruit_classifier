# Create transformers

```python
train_transform = transforms.Compose([ 
    transforms.ToTensor(),
    transforms.RandomRotation(10),
    transforms.RandomHorizontalFlip(),
    transforms.Normalize([0.485,0.456,0.406],
                         [0.229,0.224,0.225])
])
```

```python
test_transform = transforms.Compose([
    transforms.ToTensor(),
    transforms.Normalize([0.485,0.456,0.406],
                         [0.229,0.224,0.255])
])
```

```python
train_dir='fruits-360-original-size/Training/'
test_dir='fruits-360-original-size/Test/'
```

* Read & transform our data

```python
train_data = datasets.ImageFolder(os.path.join(train_dir),transform=train_transform)
test_data = datasets.ImageFolder(os.path.join(test_dir),transform=test_transform)
```

* create dataloaders

```python
train_loader=DataLoader(train_data,batch_size=2026,shuffle=True)
test_loader=DataLoader(test_data,batch_size=2026,shuffle=False)

class_names= train_data.classes
class_names
```
